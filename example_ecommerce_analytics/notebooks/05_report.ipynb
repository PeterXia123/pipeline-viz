# 05 - Report Generation
Combine segmentation with order features into final reports.

```python
import json
import pandas as pd
from src.utils.io_helpers import ensure_dir, write_json
```

```python
segments = pd.read_csv('data/models/segments.csv')
order_features = pd.read_csv('data/features/order_features.csv')
with open('data/models/centroids.json') as f:
    centroids = json.load(f)
```

```python
summary = {
    'total_users': int(len(segments)),
    'segment_centroids': centroids,
    'monthly_revenue': order_features.to_dict(orient='records'),
}
ensure_dir('reports')
write_json(summary, 'reports/summary.json')
```

```python
report = segments[['user_id', 'segment', 'total_spend_usd', 'order_count']]
report.to_csv('reports/segment_report.csv', index=False)
```
