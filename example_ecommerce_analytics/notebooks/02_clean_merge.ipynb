# 02 - Clean & Merge
Merge orders with products and users, convert currency.

```python
import pandas as pd
from src.etl.cleaners import clean_orders, merge_order_details, convert_currency
from src.utils.io_helpers import load_exchange_rate, ensure_dir
```

```python
orders = pd.read_csv('data/clean/orders_validated.csv', parse_dates=['order_date'])
products = pd.read_csv('data/clean/products_validated.csv')
users = pd.read_csv('data/clean/users_validated.csv', parse_dates=['signup_date'])
```

```python
orders_clean = clean_orders(orders)
merged = merge_order_details(orders_clean, products, users)
rate = load_exchange_rate('CNY')
merged = convert_currency(merged, rate)
```

```python
ensure_dir('data/clean')
merged.to_csv('data/clean/merged_orders.csv', index=False)
```
