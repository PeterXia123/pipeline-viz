# 01 — Ingest & merge
Merge raw tables; enrich with external geo via `src.ccds_ds.ingest`.

```python
import pandas as pd
from pathlib import Path
from src.ccds_ds.ingest import merge_with_geo

Path('data/interim').mkdir(parents=True, exist_ok=True)

customers = pd.read_csv('/Users/admin/Documents/pipeline-viz/example_cookiecutter_ds/data/raw/customers.csv')
orders = pd.read_csv('data/raw/orders.csv')
merged = orders.merge(customers, on='customer_id', how='left')
merged = merge_with_geo(merged, Path('.').resolve())
merged.to_csv('data/interim/merged_orders.csv', index=False)
```
